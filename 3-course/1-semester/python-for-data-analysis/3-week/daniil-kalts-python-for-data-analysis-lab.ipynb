### Exercise 1

Write a program that takes in a filename, then takes in a series of lines of input until a blank line is entered, writing each line to the file with the given name. After the blank line is entered, properly close the file before ending the program.

```python
filename = input("Enter filename: ")

with open(filename, "w") as op:
    while True:
        line_val = input("Enter a new line: ")

        if line_val == "":
            break

        op.write(line_val + "\n")
```

### Exercise 2

Write a program that takes in a filename and string as input. Then print how many times that string appears inside the chosen file. If the file does not exist, continue asking for a filename until one is given that exists. Use your source code file as test input.

```python
your_substring = input("Enter your string: ")

while True:
    filename = input("Enter your file name: ")

    try:
        with open(filename, "r") as file:
            content = file.read()
        break
    except FileNotFoundError:
        print("File does not exist. Try again.")

substring_count = content.count(your_substring)

print(f"The file contains {substring_count} occurrences.")
```

Task 2. The input file contains a single text string, possibly containing spaces. Print this line in reverse order. The line in the input file ends with the end of line character '\n'.

```python
with open("option1_task2_input.txt", "r") as file:
    line = file.readline().rstrip("\n")

print(line[::-1])
```

Task 1. The input file contains two integers that can be separated by spaces and line ends. Output their sum to the output file.
Note. Read the entire file into a string variable using the read() method and split it into parts using the split () method.

Task 2. Print all the lines in this file in reverse order. 
To do this, read the list of all rows using the read () lines method.
The last line of the input file must end with the character '\n'. 

```python
with open("option2_task2_input.txt", "r") as infile:
    lines = infile.readlines()

for line in reversed(lines):
    print(line, end="")
```

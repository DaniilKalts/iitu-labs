### Exercise 1

Write a Python script that lets the user do some basic geometry calculations. Your output should look something like the following. The majority of the output must be calculated by your program using a function for each kind of calculation. The output given below is just an example. Your program needs to perform the calculations for the numerical values read from the keyboard. 

```python
import math


def rectangle_area(h, w):
    return h * w


def rectangle_perimeter(h, w):
    return 2 * h + 2 * w


def rectangle_diagonal(h, w):
    return math.sqrt(h ** 2 + w ** 2)


def circle_area(r):
    return math.pi * r ** 2


def circle_circumference(r):
    return 2 * math.pi * r


def triangle_area(h, b):
    return h * b / 2


def triangle_perimeter(h, b):
    c = math.sqrt(h ** 2 + b ** 2)
    return h + b + c


def exterior_angle(n):
    return 360 / n


def interior_sum(n):
    return (n - 2) * 180


def one_interior_angle(n):
    return interior_sum(n) / n


def polygon_area(n, side):
    return n * side ** 2 / (4 * math.tan(math.pi / n))


h = float(input("Enter height of a rectangle >>> "))
w = float(input("Enter width of a rectangle >>> "))

print()
print("The area of a rectangle with height", h, "and width", w,
      "is", rectangle_area(h, w), ".")
print("The perimeter of a rectangle with height", h, "and width", w,
      "is", rectangle_perimeter(h, w), ".")
print("The length of the diagonal of a rectangle with height", h,
      "and width", w, "is", rectangle_diagonal(h, w), ".")


r = float(input("\nEnter the radius of a circle >>> "))

print()
print("The area of a circle with radius", format(r, ".2f"),
      "is", format(circle_area(r), ".2f"), ".")
print("The circumference of a circle with radius", format(r, ".2f"),
      "is", format(circle_circumference(r), ".2f"), ".")


th = float(input("\nEnter the height of a right triangle >>> "))
tb = float(input("Enter the base of a right triangle >>> "))

print()
print("The area of a right triangle with height", format(th, ".2f"),
      "and base", format(tb, ".2f"), "is",
      format(triangle_area(th, tb), ".2f"), ".")

print("The perimeter of a triangle with height", format(th, ".2f"),
      "and base", format(tb, ".2f"), "is",
      format(triangle_perimeter(th, tb), ".2f"), ".")


n = int(input("\nEnter the number of sides of a regular polygon as >>> "))
side = float(input("Enter the length of the side of a regular polygon >>> "))

print()
print("The exterior angle of a regular polygon with", n,
      "sides is", format(exterior_angle(n), ".2f"), "degrees.")

print("The interior angles of a regular polygon with", n,
      "sides sum to", format(interior_sum(n), ".2f"), "degrees.")

print("Each interior angle of a regular polygon with", n,
      "sides is", format(one_interior_angle(n), ".2f"), "degrees.")

print("The area of a regular polygon with", n,
      "sides each", format(side, ".2f"), "long is",
      format(polygon_area(n, side), ".2f"), ".")
```

### Exercise 2
You buy an international calling card to Germany. The calling card company has some special offers.
(a) If you charge your card with less than $ 10, you don’t get anything extra.
(b) For a less than $ 25 charge, you get $ 3 of extra phone time.
(c) For a less than $ 50 charge, you get $ 8 of extra phone time.
(d) For a less than $ 100 charge, you get $ 20 of extra phone time.
(e) For a more than $ 100 charge, you get $ 25 of extra phone time.
Write a function that takes the value the user wants to charge and returns the actual value charged.
In your script, include a way for someone running the script to enter values to charge and get the actual value charged.

```python
def actual_charge(amount):
    if amount < 10:
        return amount
    elif amount < 25:
        return amount + 3
    elif amount < 50:
        return amount + 8
    elif amount <= 100:
        return amount + 20
    else:
        return amount + 25


charge = float(input("Enter value you want to charge >>> "))

result = actual_charge(charge)

print(result, "dollars were added.")
```

```python
import pandas as pd
```

```python
marksUT= {'Name':['Alvin','Alvin','Alvin','Belle','Belle','Belle', 'Chip','Chip','Chip','Dale','Dale','Dale'],
'UT':[1,2,3,1,2,3,1,2,3,1,2,3], #Unit test 1,2,3
'Math':[22,21,14,20,23,22,23,24,12,15,18,17],
'Science':[21,20,19,17,15,18,19,22,25,22,21,18],
'Engl':[18,17,15,22,21,19,20,24,19,25,25,20],
'Hist':[20,22,24,24,25,23,15,17,21,22,24,25],
'Liter':[21,24,23,19,15,13,22,21,23,22,23,20]
}
df=pd.DataFrame(marksUT)
print(df)
```

# 2. Data Agregation

```python
df.aggregate('max')
```

```python
#To use multiple aggregate functions in a single statement
df.aggregate(['max','count','min'])
```

```python
df['Math'].aggregate(['max','min'])
```

```python
#Using the above statement with axis=0 gives the same result
df['Math'].aggregate(['max','min'],axis=0)
```

```python
#Total marks of Maths and Science obtained by each student.
#Use sum() with axis=1 (Row-wise summation)
df[['Math','Science']].aggregate('sum',axis=1)
```

# Working with Delimited Formats

```python
import csv
f = open(path+'ex7.csv')
reader = csv.reader(f)
```

```python
for line in reader:
    print(line)
```

```python
with open(path+'ex7.csv') as f:
    lines = list(csv.reader(f))
    
```

```python
header, values = lines[0], lines[1:]
```

```python
data_dict = {h: v for h, v in zip(header, zip(*values))}
data_dict
```

```python
class my_dialect(csv.Dialect):
    lineterminator = '\n'
    delimiter = ';'
    quotechar = '"'
    quoting = csv.QUOTE_MINIMAL
```

```python
with open(path+'ex7.csv') as f:
    reader = csv.reader(f, dialect=my_dialect)
    reader = csv.reader(f, delimiter='|')
```

```python
with open('mydata.csv', 'w') as f:
    writer = csv.writer(f, dialect=my_dialect)
    writer.writerow(('one', 'two', 'three'))
    writer.writerow(('1', '2', '3'))
    writer.writerow(('4', '5', '6'))
    writer.writerow(('7', '8', '9'))
```

# Parsing XML with lxml.objectify

## Extensible Markup Language (XML)
- Similar to HTML, but designed for data storage/transport.
- Uses tags and attributes to structure information.

```python
from lxml import objectify
path2 = '/Users/alimbekov/Desktop/Gaukhar/2025-2026/Python for Data Analysis/practice/datasets/'
parsed = objectify.parse(open(path2+'mta_perf/Performance_MNR.xml'))
root = parsed.getroot()
```

```python
data = []
skip_fields = ['PARENT_SEQ', 'INDICATOR_SEQ', 'DESIRED_CHANGE', 'DECIMAL_PLACES']
```

```python
for elt in root.INDICATOR:
    el_data = {}
    for child in elt.getchildren():
        if child.tag in skip_fields:
            continue
        el_data[child.tag] = child.pyval
    data.append(el_data)
```

```python
perf = pd.DataFrame(data)
perf.head()
```

# Binary Data Formats

One of the easiest ways to store data (also known as serialization) efficiently in binary
format is using Python’s built-in pickle serialization. pandas objects all have a
to_pickle method that writes the data to disk in pickle format:

Pickle in Python is a module (pickle) that allows you to serialize (convert an object into a byte stream) and deserialize (reconstruct the object from that byte stream).

## Binary Formats (Efficient Storage)
### a) Pickle
- Python-specific serialization.
- Fast, preserves Python objects exactly.
- Not cross-language safe.

### Parquet / Feather
- Columnar storage → very efficient for big data.
- Great for analytics, supported by Spark, Arrow, etc.

# Using HDF5 Format

### Hierarchical Data Format v5 (HDF5).
Binary format for large, complex, structured datasets.
Supports:
- Storing multiple datasets in one file (like sheets in Excel).
- Compression (saves space).
- Random access (read part of a dataset without loading all).

HDF5 is very efficient for large datasets (faster than CSV, smaller files).
Good for incremental storage (append new data without rewriting the whole file).

```python
frame = pd.DataFrame({'a': np.random.randn(100)})
```

```python
store = pd.HDFStore('mydata.h5')
```

```python
store['obj1'] = frame
```

```python
store['obj1_col'] = frame['a']
```

```python
store
```

```python
store['obj1']
```

```python
store.put('obj2', frame, format='table')
store.select('obj2', where=['index >= 10 and index <= 15'])
store.close()
```

```python
frame.to_hdf('mydata.h5', 'obj3', format='table')
pd.read_hdf('mydata.h5', 'obj3', where=['index < 5'])
```

# Interacting with Databases

### SQL Databases
- Data is often stored in relational databases.
- Pandas can read/write via SQLAlchemy.

```python
import sqlite3
```

```python
query = """ CREATE TABLE test
(a VARCHAR(20), 
b VARCHAR(20),
c REAL,
d INTEGER
);"""
```

```python
con = sqlite3.connect('mydata4.sqlite')
# con.execute(query)
con.commit()
```

```python
data = [('Atlanta', 'Georgia', 1.25, 6),
('Tallahassee', 'Florida', 2.6, 3),
('Sacramento', 'California', 1.7, 5)]
```

```python
stmt = "INSERT INTO test VALUES(?, ?, ?, ?)"
con.executemany(stmt, data)
```

```python
con.commit() #the rows are in the database, but they aren’t permanently saved until you run:
```

```python
cursor = con.execute('select * from test')
rows = cursor.fetchall()
rows
```

```python
cursor.description
pd.DataFrame(rows, columns=[x[0] for x in cursor.description])
```

### 1. Assuming the given table: Product. Write the python code for the following: 
a) To create the data frame for the above table.

b) To add the new rows in the data frame. (2 - 3 rows as you wish) 

c) To display the maximum price of LG TV.

d) To display the Sum of all products.

e) To display the median of the USD of Sony products.

```python
import pandas as pd

# a)
df = pd.DataFrame({
    "Item": ["TV", "TV", "TV", "AC"],
    "Company": ["LG", "VIDEOCON", "LG", "SONY"],
    "Peso": [12000, 10000, 15000, 14000],
    "USD": [700, 650, 800, 750]
})

print("a) Original DataFrame:")
print(df)
print()

# b)
new_rows = pd.DataFrame({
    "Item": ["Smartphone", "Smartphone", "TV"],
    "Company": ["Samsung", "Apple", "SONY"],
    "Peso": [8000, 12000, 18000],
    "USD": [500, 900, 1100]
})

df = pd.concat([df, new_rows], ignore_index=True)

print("b) DataFrame after adding new rows:")
print(df)
print()

# c)
lg_tvs = df[(df["Company"] == "LG") & (df["Item"] == "TV")]

print("c) Maximum price of an LG TV:")
print(lg_tvs[["Peso", "USD"]].max())
print()

# d)
print("d) Sum of all products:")
print(df[["Peso", "USD"]].sum())
print()

# e)
sony_products = df[df["Company"] == "SONY"]
median_sony_usd = sony_products["USD"].median()

print("e) Median USD price of Sony products:")
print(median_sony_usd)
```

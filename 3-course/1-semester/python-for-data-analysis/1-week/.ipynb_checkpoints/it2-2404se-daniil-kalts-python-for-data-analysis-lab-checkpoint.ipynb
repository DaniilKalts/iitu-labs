### Exercise 1 : Squares of Numbers

```python
"""
Exercise 1: Squares of Numbers
Given the list:
numbers = [1, 2, 3, 4, 5]
Use a list comprehension to create a list of squares.
"""

numbers = [1, 2, 3, 4, 5]
numbers_squared = [x * x for x in numbers]

print(numbers)
print(numbers_squared)
```

### Exercise 2: Even Numbers

```python
"""
Given
numbers = list(range(1, 21))
Create a list containing only even numbers using list comprehension.
"""

numbers = list(range(1, 21))
even_numbers = [x for x in numbers if x % 2 == 0]

print(numbers)
print(even_numbers)
```

### Exercise 3: Convert Strings to Uppercase

```python
"""
Given
words = ["python", "java", "c++", "javascript"]
Create a new list where every word is uppercase.
"""

words = ["python", "java", "c++", "javascript"]
capitalized_words = [word.capitalize() for word in words]

print(words)
print(capitalized_words)
```

### Exercise 4: Word Lengths

```python
"""
Given
words = ["apple", "banana", "grape", "kiwi"]
Create a list containing the length of each word.
"""

words = ["apple", "banana", "grape", "kiwi"]
words_lengths = [len(word) for word in words]

print(words)
print(words_lengths)
```

### Exercise 5: Create a Dictionary

```python
"""
Given
students = ["Alice", "Bob", "Charlie", "David"]
Create a dictionary where each student has an initial score of 0.
"""

students = ["Alice", "Bob", "Charlie", "David"]
student_score = {student: 0 for student in students}

print(students)
print(student_score)
```

### Exercise 6: Unique Characters

```python
"""
Given
text = "programming"
Create a set containing all unique letters.
"""

text = "programming"
unique_text = {letter for letter in text}

print(text)
print(unique_text)
```

### Exercise 7: Flatten a Nested List

```python
"""
Given
matrix = [
    [1, 2, 3],
    [4, 5],
    [6, 7, 8]
]
Use a nested list comprehension to create a single flat list.
"""

matrix = [
    [1, 2, 3],
    [4, 5],
    [6, 7, 8]
]
flat_matrix = [item for sublist in matrix for item in sublist]

print(matrix)
print(flat_matrix)
```

### Challenge 1: Prime Numbers

```python
"""
Using list comprehension, generate all prime numbers from 2 to 100.
"""

primes = [x for x in range(2, 101) if all(x % i != 0 for i in range(2, int(x**0.5) + 1))]

print(primes)
```

### Challenge 2: Student Grades

```python
"""
grades = {
    "Alice": 82,
    "Bob": 59,
    "Charlie": 91,
    "David": 74,
    "Emma": 98
}
Using dictionary comprehension, create a new dictionary containing only students with grades 80 or higher.
"""

grades = {
    "Alice": 82,
    "Bob": 59,
    "Charlie": 91,
    "David": 74,
    "Emma": 98
}

cool_guys = {guy: grade for guy, grade in grades.items() if grade >= 80}

print(grades)
print(cool_guys)
```

### Challenge 3: Number Classification

```python
"""
Given

numbers = range(-10, 11)

Using a list comprehension with conditional expressions, create a list where:
negative numbers become "Negative"
zero becomes "Zero"
positive numbers become "Positive"

Expected Output
[
'Negative',
'Negative',
...
'Zero',
'Positive',
'Positive'
]
"""

numbers = range(-10, 11)
verbal_numbers = ["Negative" if x < 0 else "Zero" if x == 0 else "Positive" for x in numbers]

print(list(numbers))
print(verbal_numbers)
```

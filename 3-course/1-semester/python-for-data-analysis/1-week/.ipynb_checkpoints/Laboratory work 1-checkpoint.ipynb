# List Comprehension and Other Comprehensions

Brief Explanation

## 1. List Comprehension
A list comprehension is a concise way to create a new list from an existing iterable.

```python
numbers = [1, 2, 3, 4, 5]
squares = [x ** 2 for x in numbers]

print(squares)
```

## 2. Set Comprehension
A set comprehension creates a set. Sets automatically remove duplicate values.

```python
numbers = [1, 2, 2, 3, 3, 4]

unique = {x for x in numbers}

print(unique)
```

## 3. Dictionary Comprehension
A dictionary comprehension creates dictionaries quickly.

```python
numbers = [1, 2, 3, 4]

square_dict = {x: x ** 2 for x in numbers}

print(square_dict)
```

## 4. Generator Expression
A generator expression produces values one at a time instead of storing them all in memory.

```python
numbers = (x ** 2 for x in range(5))

for num in numbers:
    print(num)
```

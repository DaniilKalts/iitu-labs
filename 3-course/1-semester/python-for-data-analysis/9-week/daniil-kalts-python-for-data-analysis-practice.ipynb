### Exercise 2 – Average Purchase Amount
Write a statement to find the **average purchase amount** of all orders.

### Exercise 3 – Number of Salesmen
Write a statement to find the **number of salesmen** currently listing for all of their customers (count the **unique** `salesman_id` values).

**Expected result:** `6`

```python
cnt = orders['salesman_id'].drop_duplicates().count()
print('Number of salesman:', cnt)
```

### Exercise 4 – Maximum Purchase Amount

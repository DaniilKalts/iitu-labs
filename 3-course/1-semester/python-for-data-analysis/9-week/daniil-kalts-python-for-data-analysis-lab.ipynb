### Exercise 1 – Maximum Price
Find the **maximum price** of any item ordered in the **items_ordered** table.

**Expected result:** `1250.00`

````python
import datetime

def fix_price(value):
  if isinstance(value, datetime.datetime):        
      return value.month + (value.year % 100) / 100 
  return value                     

items['price'] = pd.to_numeric(items['price'].apply(fix_price))

max_price = items['price'].max()
print('Maximum price:', max_price)
````

### Exercise 8. Task 1 – Plane Holidays in Resorts Starting with "S"
Create a **pivot table** from this data, then use the filters (slice) within it to view the **average prices** of holidays that have:
- a **Travel Method** of **Plane**, and
- a **Resort Name** that begins with the letter **S** (**Seville** and **Santiago**).

````python
holidays_pivot = pd.pivot_table(
    list_of_holidays,
    index=['Travel Method', 'Resort Name'],
    values='Price',
    aggfunc='mean'
)

plane_s = holidays_pivot.loc['Plane']
plane_s[plane_s.index.str.startswith('S')]
````

### Exercise 9 – Life is a Rollercoaster
Download the file **"Life is a Rollercoaster.xlsx"**. Convert this data into a **pivot table** and find the **overall average speed** of all rides that satisfy **all** of the following criteria:
- The **Type** is **Steel**;
- The **Design** is **Sit Down**;
- And they **Opened after 2000**.

````python
coasters = pd.read_excel('datasets/Life is a Rollercoaster.xlsx')

new_coasters = coasters[coasters['Opened'] > 2000]

pd.pivot_table(
    new_coasters,
    index=['Type', 'Design'],
    values='Speed ( mph )',
    aggfunc='mean'
).loc[('Steel', 'Sit Down')]
````

### Exercise 10. Task 1 – Toyota, Nissan, Mazda, Honda, Mitsubishi, Subaru, Volkswagen, Volvo
Shorten the table so it only includes the 8 makes listed:
`["toyota", "nissan", "mazda", "honda", "mitsubishi", "subaru", "volkswagen", "volvo"]`

**Tasks:**
- a. Use **`pd.crosstab`** to look at how many cars of each **body style** these car makers made.
- b. Calculate the **average curb weight** of cars by **body style** and **manufacturer**.

````python
makes = ["toyota", "nissan", "mazda", "honda", "mitsubishi", "subaru", "volkswagen", "volvo"]
df = df_raw[df_raw['make'].isin(makes)]

display(pd.crosstab(df['make'], df['body_style']))

pd.pivot_table(df, index='make', columns='body_style', values='curb_weight', aggfunc='mean')
````

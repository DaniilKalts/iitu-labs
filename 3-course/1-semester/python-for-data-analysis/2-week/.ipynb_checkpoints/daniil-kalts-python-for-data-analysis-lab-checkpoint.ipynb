### Exercise 1
Any fraction can be written as the division of two integers. You could express this in Python as a tuple – (numerator, denominator). Write functions for each of the following. They must use the tuple representation to return fractions.

1. Given two fractions as tuples, multiply them.
2. Given two fractions as tuples, divide them.
3. Given a list of fractions as a tuple, return the one that is smallest in value. Also write a small command-line interface such that the user running your script sees something like this:

```python
def multiply_fractions(fraction_1, fraction_2):
    return (
        fraction_1[0] * fraction_2[0],
        fraction_1[1] * fraction_2[1]
    )


def divide_fractions(fraction_1, fraction_2):
    if fraction_1[1] == fraction_2[1]:
        return (
            fraction_1[0],
            fraction_2[0]
        )

    return (
        fraction_1[0] * fraction_2[1],
        fraction_1[1] * fraction_2[0]
    )


def smallest_fraction(fractions):
    smallest = fractions[0]

    for fraction in fractions:
        if fraction[0] / fraction[1] < smallest[0] / smallest[1]:
            smallest = fraction

    return smallest


print('== Multiplication and Division ==')

fraction_1 = tuple(int(part) for part in input('Enter a fraction >>> ').split('/'))
fraction_2 = tuple(int(part) for part in input('Enter a fraction >>> ').split('/'))

multiply_fractions_out = multiply_fractions(fraction_1, fraction_2)
divide_fractions_out = divide_fractions(fraction_1, fraction_2)

print(
    'Multiplication of the fractions:',
    f'{multiply_fractions_out[0]}/{multiply_fractions_out[1]}'
)
print(
    'Division of the first by the second:',
    f'{divide_fractions_out[0]}/{divide_fractions_out[1]}'
)

print()
print('== Smallest fraction ==')

list_fractions = []

while True:
    val = input('Enter a fraction >>> ')

    if val == 'stop':
        break

    list_fractions.append(tuple(int(part) for part in val.split('/')))

smallest_fraction_out = smallest_fraction(list_fractions)

print(
    'Smallest fraction:',
    f'{smallest_fraction_out[0]}/{smallest_fraction_out[1]}'
)
```

### Exercise 2
Take in numbers as input until “stop” is entered. Then split the numbers into three lists: one containing all the numbers, one containing all even values, and one containing all odd. Print out all three lists, as well as each list’s sum and average. Assume all input values are integers. Sample: 

```python
def print_list_info(name, numbers):
    print(f'{name} numbers:', numbers)

    if len(numbers) > 0:
        print(f'Average of {name.lower()} numbers:', sum(numbers) / len(numbers))
    else:
        print(f'Average of {name.lower()} numbers:', 0)

    print(f'Sum of {name.lower()} numbers:', sum(numbers))


all_numbers = []
even_numbers = []
odd_numbers = []

while True:
    val = input('Input a number >>> ')

    if val == 'stop':
        break

    number = int(val)

    all_numbers.append(number)

    if number % 2 == 0:
        even_numbers.append(number)
    else:
        odd_numbers.append(number)


print_list_info('All', all_numbers)
print_list_info('Even', even_numbers)
print_list_info('Odd', odd_numbers)
```

### Exercise 3
Take in numbers as input until “stop” is entered. As you take in each number, insert it into a list so that the list is sorted in ascending order. That is, look through the list until you find the place where the new element belongs, then use .insert() to place it there. If the number isalready in the list, do not add it again. After “stop” is entered, print out the list. Do not use any of Python’s built-in sorting functions. You cannot use .sort() for this exercise. Sample: 

```python
numbers = []

while True:
    val = input('Input a number >>> ')

    if val == 'stop':
        break

    number = float(val)

    if number not in numbers:
        index = 0

        while index < len(numbers) and numbers[index] < number:
            index += 1

        numbers.insert(index, number)

    print('List contains', numbers)

print(numbers)
```

### Exercise 4
Task: Student Performance Analyzer
Create a function analyze_students() that takes three lists: student names, math scores, and Python scores.
Use zip() to combine the data and enumerate() to assign each student a number starting from 1.
Return a dictionary containing each student’s name, scores, average score, and performance status:
Excellent: average ≥ 90
Good: average ≥ 75
Needs Improvement: average < 75\

```python
def analyze_students(students, math_scores, python_scores):
    student_data = {}

    for number, (student, math_score, python_score) in enumerate(
        zip(students, math_scores, python_scores),
        start=1
    ):
        average = (math_score + python_score) / 2

        if average >= 90:
            status = 'Excellent'
        elif average >= 75:
            status = 'Good'
        else:
            status = 'Needs Improvement'

        student_data[number] = {
            'name': student,
            'math': math_score,
            'python': python_score,
            'average': average,
            'status': status
        }

    return student_data


students = ['Ali', 'Bob', 'Charlie', 'Diana', 'Ethan']

math_scores = [78, 92, 65, 88, 92]

python_scores = [85, 89, 70, 91, 95]

result = analyze_students(students, math_scores, python_scores)

print(result)
```

### Option 1.

Task 1. Find the number of positive (>0) items in this list.

Task 2. A list of numbers is given. Print the value of the largest item in the list, and then the index of that item in the list. If there are several largest elements, output the index of the first one.

Task 3. Print the value of the smallest odd element in the list, and if there are no odd elements in the list, print the number 0.

```python
# Task 1.

numbers = [3, -5, 0, 12, -1, 7]

positive_count = 0

for number in numbers:
    if number > 0:
        positive_count += 1

print('Task 1:')
print(positive_count)
print()

# Task 2.

numbers = [3, 12, -5, 12, 7]

largest = numbers[0]
largest_index = 0

for index, number in enumerate(numbers):
    if number > largest:
        largest = number
        largest_index = index

print('Task 2:')
print(largest)
print(largest_index)
print()

# Task 3.

numbers = [8, 12, 4, 7, 3, 10, 9]

smallest_odd = None

for number in numbers:
    if number % 2 != 0:
        if smallest_odd is None or number < smallest_odd:
            smallest_odd = number

print('Task 3:')
if smallest_odd is None:
    print(0)
else:
    print(smallest_odd)
print()
```

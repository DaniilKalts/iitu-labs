# 4. Use the following dataset for all exercises 

1) Find all unique departments in the dataset.

2) How many times does each Name appear?

3) Check whether "Alice" and "David" are present in the Name column.

4) Compute the mean, median, and standard deviation of the Salary column.

5) Get summary statistics (.describe()) for the numeric columns.

6) Find the maximum salary in each department.

7) Notice that "Alice" and "Bob" appear multiple times. Extract all rows for "Alice".

9) How does having duplicate names affect indexing?

10) Sort employees by Salary in descending order.

11) Sort first by Department and then by Joining_Year.

12) Rank employees by their Bonus (highest bonus = rank 1).

13) Create a new column Salary_in_Thousands using .apply() to divide salary by 1000.

14) Use .map() to replace department names with abbreviations: {"Finance": "FIN", "HR": "HR", "IT": "IT"}.

15) Use .apply() with a custom lambda function to classify employees as "Senior" if Joining_Year <= 2016 else "Junior".

16) Create a Series with average salaries per department and subtract it from each employee’s salary.

17) Add a Series with extra yearly allowance = [1000, 2000, ..., 10000] to the Bonus column. Compare the results with direct broadcasting.

18) Add Salary and Bonus, but replace missing Bonus values with 0.

19) Multiply Salary by a factor Series (e.g., pd.Series([1.05, 1.1, 1.0], index=["HR", "Finance", "IT"])), using .add() or .mul() with fill_value.



```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "Alice", "Eve", "Frank", "Grace", "Bob", "Hannah", "Ian"],
    "Department": ["HR", "Finance", "IT", "HR", "Finance", "IT", "Finance", "HR", "IT", "Finance"],
    "Salary": [50000, 60000, 70000, 52000, 62000, 68000, 61000, 59000, 72000, 60000],
    "Bonus": [5000, 7000, 6000, 5200, 6500, np.nan, 6100, 5800, 7100, 6900],
    "Joining_Year": [2015, 2016, 2017, 2015, 2018, 2017, 2016, 2016, 2015, 2018]
}

df = pd.DataFrame(data)
print(df)
print()

# 1)
print("1) Unique departments:")
print(df["Department"].unique())
print()

# 2)
print("2) Name counts:")
print(df["Name"].value_counts())
print()

# 3)
print("3) Name checks:")
print("Alice:", "Alice" in df["Name"].values)
print("David:", "David" in df["Name"].values)
print()

# 4)
print("4) Salary statistics:")
print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())
print("Standard deviation:", df["Salary"].std())
print()

# 5)
print("5) Summary statistics:")
print(df.describe())
print()

# 6)
print("6) Maximum salary by department:")
print(df.groupby("Department")["Salary"].max())
print()

# 7)
print("7) Rows for Alice:")
print(df[df["Name"] == "Alice"])
print()

# 8)
print("Duplicate names do not affect the default index; each occurrence is in a separate row.")
print()

# 9)
print("9) Employees sorted by salary:")
print(df.sort_values(by="Salary", ascending=False))
print()

# 10)
print("10) Employees sorted by department and joining year:")
print(df.sort_values(by=["Department", "Joining_Year"]))
print()

# 11)
df["Bonus_Rank"] = df["Bonus"].rank(ascending=False)

print("11) Employees ranked by bonus:")
print(df[["Name", "Bonus", "Bonus_Rank"]].sort_values(by="Bonus_Rank"))
print()

# 12)
df["Salary_in_Thousands"] = df["Salary"].apply(lambda salary: salary / 1000)

print("12) Salaries in thousands:")
print(df)
print()

# 13)
df["Department"] = df["Department"].map({
    "Finance": "FIN",
    "HR": "HR",
    "IT": "IT"
})

print("13) Departments with abbreviations:")
print(df)
print()

# 14)
df["Seniority"] = df["Joining_Year"].apply(
    lambda joining_year: "Senior" if joining_year <= 2016 else "Junior"
)

print("14) Employee seniority:")
print(df[["Name", "Joining_Year", "Seniority"]])
print()

# 15)
avg_salaries_per_department = df.groupby("Department")["Salary"].mean()

df["Salary_minus_Department_Avg"] = (
    df["Salary"] - df["Department"].map(avg_salaries_per_department)
)

print("15) Salary compared with department average:")
print(df[["Name", "Department", "Salary", "Salary_minus_Department_Avg"]])
print()

# 16)
extra_allowance = pd.Series([i * 1000 for i in range(1, 11)])

print("Bonus + Series:")
print(df["Bonus"] + extra_allowance)
print()

print("Bonus + list:")
print(df["Bonus"] + [i * 1000 for i in range(1, 11)])
print()

# 17)
print("17) Salary + Bonus (missing bonuses treated as 0):")
print(df["Salary"] + df["Bonus"].fillna(0))
print()

# 18)
department_factors = pd.Series(
    [1.05, 1.1, 1.0],
    index=["HR", "FIN", "IT"]
)

df["Adjusted_Salary"] = df["Salary"].mul(
    df["Department"].map(department_factors),
    fill_value=0
)

print("18) Salary multiplied by department factor:")
print(df[["Name", "Department", "Salary", "Adjusted_Salary"]])
print()
```

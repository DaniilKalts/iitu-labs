## Operations on Arrays 

### Arithmetic Operations  

```python
array1 = np.array([[3,6],[4,2]])
array2 = np.array([[10,20],[15,12]])
```

```python
#Element-wise addition of two matrices.
array1 + array2
```

```python
#Subtraction
array1 - array2
```

```python
#Multiplication
array1 * array2
```

```python
#Matrix Multiplication
array1 @ array2
```

```python
#Exponentiation
array1 ** 3
```

```python
#Division
array2 / array1
```

```python
#Element wise Remainder of Division
#(Modulo)
array2 % array1
```

### Transpose

```python
#Transpose
array3 = np.array([[10,-7,0, 20],[-5,1,200,40],[30,1,-1,4]])
array3
```

```python
# the original array does not change
array3.transpose() 
```

### Sorting  

```python
array4 = np.array([1,0,2,-3,6,8,4,7])
array4.sort()
array4
```

```python
array4 = np.array([[10,-7,0, 20],[-5,1,200,40],[30,1,-1,4]])
array4
```

```python
#default is row-wise sorting
array4.sort() 
array4
```

```python
array5 = np.array([[10,-7,0, 20], [-5,1,200,40],[30,1,-1,4]])
#axis =0 means column-wise sorting 
array5.sort(axis=0)
array5
```

## Splitting Arrays

```python
array4 = np.array([[ 10, -7, 0, 20],
 [ -5, 1, 200, 40],
 [ 30, 1, -1, 4],
 [ 1, 2, 0, 4],
 [ 0, 1, 0, 2]])
```

```python
array4
```

```python
# [1,3] indicate the row indices on which
# to split the array 
first, second, third = np.split(array4, [1, 3])
```

```python
# array4 is split on the first row and
# stored on the sub-array first 
first
```

```python
# array4 is split after the first row and
# upto the third row and stored on the
# sub-array second
second
```

```python
# the remaining rows of array4 are stored
# on the sub-array third
third
```

```python
#[1, 2], axis=1 give the columns indices
#along which to split
firstc, secondc, thirdc =np.split(array4,[1, 2], axis=1)
```

```python
firstc
```

```python
secondc
```

```python
thirdc
```

```python
# 2nd parameter 2 implies array is to be
# split in 2 equal parts axis=1 along the
# column axis
firsthalf, secondhalf =np.split(array4,2,axis=1)
```

```python
firsthalf
```

```python
secondhalf
```

## Statistical Operations on Arrays

```python
arrayA = np.array([1,0,2,-3,6,8,4,7])
arrayB = np.array([[3,6],[4,2]])
```

### max()

```python
# max element form the whole 1-D array
arrayA.max() 
```

```python
# max element form the whole 2-D array
arrayB.max()
```

```python
# if axis=1, it gives column wise maximum
arrayB.max(axis=1) 
```

```python
#if axis=0, it gives row wise maximum 
arrayB.max(axis=0) 
```

### min()

```python
arrayA.min()
```

```python
arrayB.min()
```

```python
arrayB.min(axis=0)
```

### sum() 

```python
arrayA.sum()
```

```python
arrayB.sum()
```

```python
#axis is used to specify the dimension
#on which sum is to be made. Here axis = 1
#means the sum of elements on the first row 
arrayB.sum(axis=1)
```

### mean()

```python
arrayA.mean()
```

```python
arrayB.mean()
```

```python
arrayB.mean(axis=0)
```

```python
arrayB.mean(axis=1)
```

### std() 

```python
arrayA.std()
```

```python
arrayB.std()
```

```python
arrayB.std(axis=0)
```

```python
arrayB.std(axis=1)
```

## Loading Arrays from Files 

### Using NumPy.loadtxt() 

```python
studentdata = np.loadtxt('data.txt', skiprows=1, delimiter=',',dtype = int)
```

```python
studentdata
```

```python
# To import data into multiple NumPy arrays
# row wise. Values related to student1 in
# array stud1, student2 in array stud2 etc.
stud1, stud2, stud3, stud4, stud5 = np.loadtxt('data.txt',skiprows=1,delimiter=',', dtype = int)
```

```python
stud1
```

```python
stud2
```

```python
# Import data into multiple arrays column
# wise. Data in column RollNo will be put
# in array rollno, data in column Marks1
# will be put in array mks1 and so on.
rollno, mks1, mks2, mks3 =np.loadtxt('data.txt',skiprows=1, delimiter=',', unpack=True,dtype = int)
```

```python
rollno
```

```python
mks1
```

```python
mks2
```

```python
mks3
```

###  Using NumPy.genfromtxt()

```python
dataarray = np.genfromtxt('data_w_miss.txt',skip_header=1, delimiter = ',')
```

```python
dataarray
```

```python
dataarray = np.genfromtxt('data_w_miss.txt',skip_header=1,delimiter=',', filling_values=-999,dtype = int)
```

```python
dataarray
```

```python
np.savetxt('testout.txt',studentdata, delimiter=',', fmt='%i')
```

```python
import numpy as np
```

### 1. Create the following NumPy arrays:

#### a) A 1-D array called _zeros_ having 10 elements and all the elements are set to zero.

#### b) A 1-D array called _vowels_ having the elements ‘a’, ‘e’, ‘i’, ‘o’ and ‘u’.

#### c) A 2-D array called ones having 2 rows and 5 columns and all the elements are set to 1 and dtype as int.

####  d) Use nested Python lists to create a 2-D array called _myarray1_ having 3 rows and 3 columns and store the following data:
         2.7, -2, -19
         0, 3.4, 99.9
         10.6, 0, 13
#### e) A 2-D array called _myarray2_ using _arange()_ having 3 rows and 5 columns with start value = 4, step size 4 and dtype as float.

```python
# a)
zeros = np.zeros(10)
print("a) zeros:")
print(zeros)

# b)
vowels = np.array(["a", "e", "i", "o", "u"])
print("\nb) vowels:")
print(vowels)

# c)
ones = np.ones((2, 5), dtype=int)
print("\nc) ones:")
print(ones)

# d)
myarray1 = np.array([
    [2.7, -2, -19],
    [0, 3.4, 99.9],
    [10.6, 0, 13]
])
print("\nd) myarray1:")
print(myarray1)

# e)
myarray2 = np.arange(4, 64, 4, dtype=float).reshape((3, 5))
print("\ne) myarray2:")
print(myarray2)
```

## 2. Using the arrays created in Exercise 1 above, write NumPy commands for the following:

#### a) Find the dimensions, shape, size, data type of the items and itemsize of arrays _zeros_ , _vowels_ , _ones_ , _myarray1_ and _myarray2_.

#### b) Reshape the array _ones_ to have all the 10 elements in a single row.

#### c) Display the 2nd and 3rd element of the array _vowels_.

#### d) Display all elements in the 2nd and 3rd row of the array _myarray1_.

#### e) Display the elements in the 1st and 2nd column of the array _myarray1_.

#### f) Display the elements in the 1st column of the 2nd and 3rd row of the array _myarray1_.

#### g) Reverse the array of _vowels_

```python
# a)
arrays = {
    "zeros": zeros,
    "vowels": vowels,
    "ones": ones,
    "myarray1": myarray1,
    "myarray2": myarray2
}

for name, arr in arrays.items():
    print(name)
    print("Dimensions:", arr.ndim)
    print("Shape:", arr.shape)
    print("Size:", arr.size)
    print("Data type:", arr.dtype)
    print("Item size:", arr.itemsize, "bytes")
    print()

# b)
print("b) Reshaped ones:")
print(ones.reshape((1, 10)))

# c)
print("\nc) 2nd and 3rd vowels:")
print(vowels[1:3])

# d)
print("\nd) 2nd and 3rd rows of myarray1:")
print(myarray1[1:3, :])

# e)
print("\ne) 1st and 2nd columns of myarray1:")
print(myarray1[:, 0:2])

# f)
print("\nf) 1st column of the 2nd and 3rd rows:")
print(myarray1[1:3, 0])

# g)
print("\ng) vowels in reverse order:")
print(vowels[::-1])
```

## 3. Using the arrays created in Exercise 1 from practice, write NumPy commands for the following:

####  a) Divide all elements of array _ones_ by 3.
#### b) Add the arrays _myarray1_ and _myarray2_.
#### c) Subtract _myarray1_ from _myarray2_ and store the result in a new array.
#### d) Multiply _myarray1_ and _myarray2_ elementwise.
#### e) Do the matrix multiplication of _myarray1_ and _myarray2_ and store the result in a new array _myarray3_.
#### f) Divide _myarray1_ by _myarray2_.
#### g) Find the cube of all elements of _myarray1_ and divide the resulting array by 2.
#### h) Find the square root of all elements of _myarray2_ and divide the resulting array by 2. The result should be rounded to two places of decimals.
np.around() OR .round()

```python
# Use this 3x3 array for element-wise operations
myarray2_3x3 = np.arange(4, 40, 4, dtype=float).reshape(3, 3)

# a)
print("a) ones divided by 3:")
print(ones / 3)

# b)
print("\nb) myarray1 + myarray2_3x3:")
print(myarray1 + myarray2_3x3)

# c)
new_arr = myarray2_3x3 - myarray1
print("\nc) myarray2_3x3 - myarray1:")
print(new_arr)

# d)
print("\nd) Element-wise multiplication:")
print(myarray1 * myarray2_3x3)

# e)
myarray3 = myarray1 @ myarray2
print("\ne) Matrix multiplication:")
print(myarray3)

# f)
print("\nf) myarray1 divided by myarray2_3x3:")
print(myarray1 / myarray2_3x3)

# g)
print("\ng) Cube of myarray1 divided by 2:")
print(myarray1 ** 3 / 2)

# h)
print("\nh) Square root of myarray2 divided by 2, rounded to 2 decimals:")
print(np.round(np.sqrt(myarray2) / 2, 2))
```

##  4. Create a 2-D array called _myarray4_ using _arange()_ having 14 rows and 3 columns with start value = -1, step size 0.25 having. Split this array row wise into 3 equal parts and print the result.

####    Using the _myarray4_ created in the above questions, write commands for the following:
##### a) Find the sum of all elements.
##### b) Find the sum of all elements row wise.
##### c) Find the sum of all elements column wise. Notes
##### d) Find the max of all elements.
##### e) Find the min of all elements in each row.
##### f) Find the mean of all elements in each row.
##### g) Find the standard deviation column wise.

```python
myarray4 = np.arange(-1, 9.5, 0.25).reshape(14, 3)
print(myarray4, "\n")

# a)
print("Sum of all elements:")
print(myarray4.sum())

# b)
print("\nSum of each row:")
print(myarray4.sum(axis=1))

# c)
print("\nSum of each column:")
print(myarray4.sum(axis=0))

# d)
print("\nMaximum of all elements:")
print(myarray4.max())

# e)
print("\nMinimum of each row:")
print(myarray4.min(axis=1))

# f)
print("\nMean of each row:")
print(myarray4.mean(axis=1))

# g)
print("\nStandard deviation of each column:")
print(myarray4.std(axis=0))
```

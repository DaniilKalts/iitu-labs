### Task 1. Compute the moving average of window size 3, for the given 1D array.¶

```python
import numpy as np

np.random.seed(100)

Z = np.random.randint(10, size=10)

print("Array:")
print(Z)

moving_avg_window = [Z[i:i + 3].mean() for i in range(Z.size - 2)]

print("\nMoving average:")
print(moving_avg_window)
```

### Task 2. Find the index of 5th repetition of number 1 in x.

```python
x = np.array([1, 2, 1, 1, 3, 4, 3, 1, 1, 2, 1, 1, 2])

print("Array:")
print(x)

fifth_index = np.where(x == 1)[0][4]

print("\nIndex of the 5th occurrence of 1:")
print(fifth_index)
```

### Task 3. Drop all nan values from a 1D numpy array

```python
arr = np.array([1, 2, 3, np.nan, 5, 6, 7, np.nan])

clean_arr = arr[~np.isnan(arr)]

print("Original array:")
print(arr)

print("\nArray without NaN values:")
print(clean_arr)
```

### Task 4: Weather Data Analysis with NumPy I/O

#### Data generation
Generate synthetic temperature data for 365 days of the year for 5 cities (matrix shape (365, 5)), using random numbers around realistic averages.
Save the data in three formats:

- Binary NumPy format (.npy)

- Text file (.txt)

- CSV file (.csv)

#### Data loading
Load the .npy file back into NumPy.

#### Processing
For each city, compute: 

- Minimum, maximum, mean, and standard deviation of temperature.

- The hottest and coldest day (index of the day).

- Normalize the data (convert to z-scores).

#### Advanced requirement
- Find the day of the year when the most cities had above-average temperature.

- Create a boolean mask for days when all cities were below 0°C.

#### Saving results
Save the computed statistics (per city) into a CSV file with columns: 
City, Min, Max, Mean, Std, HottestDay, ColdestDay. Name it "stats.csv"

Save the normalized data into a .npy file.

```python
import numpy as np
import csv

np.random.seed(100)

cities = ["Almaty", "Astana", "Shymkent", "Aktau", "Karaganda"]
average_temps = np.array([10, 4, 14, 13, 4])

changes = np.random.randint(-5, 6, size=(365, 5))
temps = average_temps + changes

np.save("temperatures.npy", temps)
np.savetxt("temperatures.txt", temps, fmt="%.1f")
np.savetxt(
    "temperatures.csv",
    temps,
    delimiter=",",
    fmt="%.1f",
    header="Almaty,Astana,Shymkent,Aktau,Karaganda",
    comments=""
)

temps = np.load("temperatures.npy")

minimum = temps.min(axis=0)
maximum = temps.max(axis=0)
mean = temps.mean(axis=0)
std = temps.std(axis=0)

hottest_day = temps.argmax(axis=0) + 1
coldest_day = temps.argmin(axis=0) + 1

normalized_temps = (temps - mean) / std
np.save("normalized_temperatures.npy", normalized_temps)

above_average_count = (temps > mean).sum(axis=1)
day_with_most_above_average = above_average_count.argmax() + 1

all_cities_below_zero = np.all(temps < 0, axis=1)

with open("stats.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow([
        "City", "Min", "Max", "Mean", "Std", "HottestDay", "ColdestDay"
    ])

    for i in range(len(cities)):
        writer.writerow([
            cities[i],
            round(minimum[i], 2),
            round(maximum[i], 2),
            round(mean[i], 2),
            round(std[i], 2),
            hottest_day[i],
            coldest_day[i]
        ])

print("Temperature array shape:")
print(temps.shape)

print("\nDay with the most cities above their average:")
print(day_with_most_above_average)

print("\nDays when all cities were below 0°C:")
print(np.where(all_cities_below_zero)[0] + 1)

print("\nStatistics saved to stats.csv")
print("Normalized data saved to normalized_temperatures.npy")
```

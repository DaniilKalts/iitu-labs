### Exercise 2.1 – Merging Company DataFrames
Your company operates in several cities under several managers. The DataFrames **revenue** and **managers** each hold partial information about the company, and their `city` columns don't fully match:
- The **Mendocino** branch has no revenue yet because it just opened.
- The manager of the **Springfield** branch recently left the company.

Create the DataFrames:

```python
rev = {'city': ['Austin', 'Denver', 'Springfield'], 'revenue': [100, 83, 4]}
man = {'city': ['Austin', 'Denver', 'Mendocino'], 'manager': ['Charles', 'Joel', 'Brett']}

revenue = pd.DataFrame.from_dict(rev)
managers = pd.DataFrame.from_dict(man)
```

**Tasks:**
- a. Print the DataFrames **revenue** and **managers**.
- b. **Question:** If you ran `combined = pd.merge(revenue, managers, on='city')`, how many rows would **combined** have? Run the command and explain the result.

````python
import pandas as pd

rev = {'city': ['Austin', 'Denver', 'Springfield'], 'revenue': [100, 83, 4]}
man = {'city': ['Austin', 'Denver', 'Mendocino'], 'manager': ['Charles', 'Joel', 'Brett']}

revenue = pd.DataFrame.from_dict(rev)
managers = pd.DataFrame.from_dict(man)

print(revenue)
print()
print(managers)
print()

combined = pd.merge(revenue, managers, on='city')
print(combined)
````

**Answer:** `combined` has **2 rows**. `pd.merge()` does an inner join by default, so it keeps only the cities in both tables: Austin and Denver. Springfield has no manager and Mendocino has no revenue, so both are dropped.

### Exercise 4.4 – Left & Right Merging on Multiple Columns
Besides **revenue** and **managers**, you now have a DataFrame **sales**. It summarizes units sold at specific branches, identified by `city` and `state` but not `branch_id`.

The **managers** DataFrame uses the label `branch` instead of `city`. Your task is to use left and right merges to keep all the data and find where data is missing.

- **Right merge of revenue and sales:** this shows the missing revenue values. You don't need `left_on` or `right_on`, because the merge columns have the same labels.
- **Left merge of sales and managers:** this shows the missing manager. The merge columns have different labels, so you must specify `left_on` and `right_on`.

In both cases, you're working out how to connect the fields in the rows for **Springfield**.

Create the DataFrames:

```python
rev = {'city': ['Austin', 'Denver', 'Springfield', 'Mendocino'],
       'branch_id': [10, 20, 30, 47],
       'state': ['TX','CO','IL','CA'],
       'revenue': [100, 83, 4, 200]}

man = {'branch': ['Austin', 'Denver', 'Mendocino', 'Springfield'],
       'branch_id': [10, 20, 47, 31],
       'state': ['TX','CO','CA','MO'],
       'manager': ['Charles', 'Joel', 'Brett', 'Sally']}

sale = {'city': ['Mendocino', 'Denver', 'Austin', 'Springfield', 'Springfield'],
        'state': ['CA', 'CO', 'TX', 'MO', 'IL'],
        'units': [1, 4, 2, 5, 1]}

revenue = pd.DataFrame.from_dict(rev)
managers = pd.DataFrame.from_dict(man)
sales = pd.DataFrame.from_dict(sale)
```

**Tasks:**
- a. Run a **right merge** with `pd.merge()` on **revenue** and **sales** to create a new DataFrame **revenue_and_sales**. Use `how='right'` and `on=['city', 'state']`.
- b. Print **revenue_and_sales**.
- c. Run a **left merge** on **sales** and **managers** to create a new DataFrame **sales_and_managers**. Use `how='left'`, `left_on=['city', 'state']` and `right_on=['branch', 'state']`.
- d. Print **sales_and_managers**.

````python
rev = {'city': ['Austin', 'Denver', 'Springfield', 'Mendocino'],
       'branch_id': [10, 20, 30, 47],
       'state': ['TX','CO','IL','CA'],
       'revenue': [100, 83, 4, 200]}

man = {'branch': ['Austin', 'Denver', 'Mendocino', 'Springfield'],
       'branch_id': [10, 20, 47, 31],
       'state': ['TX','CO','CA','MO'],
       'manager': ['Charles', 'Joel', 'Brett', 'Sally']}

sale = {'city': ['Mendocino', 'Denver', 'Austin', 'Springfield', 'Springfield'],
        'state': ['CA', 'CO', 'TX', 'MO', 'IL'],
        'units': [1, 4, 2, 5, 1]}

revenue = pd.DataFrame.from_dict(rev)
managers = pd.DataFrame.from_dict(man)
sales = pd.DataFrame.from_dict(sale)

revenue_and_sales = pd.merge(revenue, sales, how='right', on=['city', 'state'])
print(revenue_and_sales)
print()

sales_and_managers = pd.merge(sales, managers, how='left', left_on=['city', 'state'], right_on=['branch', 'state'])
print(sales_and_managers)
````

### Exercise 6.1 – Using merge_ordered()
The DataFrames **austin** and **houston** contain weather data for the cities of Austin and Houston.

The weather was recorded on different days, and you need to merge the two DataFrames so that the dates are in order. Use **`pd.merge_ordered()`** for this. When you're done, compare the order of the rows before and after merging.

Create the DataFrames:

```python
austin = pd.DataFrame.from_dict({'date': ['2016-01-01', '2016-02-08', '2016-01-17'],
                                 'ratings': ['Cloudy', 'Cloudy', 'Sunny']})
houston = pd.DataFrame.from_dict({'date': ['2016-01-04', '2016-01-01', '2016-03-01'],
                                  'ratings': ['Rainy', 'Cloudy', 'Sunny']})
```

**Tasks:**
- a. Run an ordered merge on **austin** and **houston** with `pd.merge_ordered()`. Store the result as **tx_weather**.
- b. Print **tx_weather**. The rows are sorted by date, but you can't tell which observation came from which city.
- c. Run another ordered merge on **austin** and **houston**, this time with `on='date'` and `suffixes=['_aus', '_hus']` so the cities can be told apart. Store the result as **tx_weather_suff**.
- d. Print **tx_weather_suff**.
- e. Run a third ordered merge on **austin** and **houston**. In addition to `on` and `suffixes`, specify `fill_method='ffill'` to replace NaN entries with the most recent non-null value. Store the result as **tx_weather_ffill**.
- f. Print **tx_weather_ffill**.

````python
austin = pd.DataFrame.from_dict({'date': ['2016-01-01', '2016-02-08', '2016-01-17'],
                                 'ratings': ['Cloudy', 'Cloudy', 'Sunny']})
houston = pd.DataFrame.from_dict({'date': ['2016-01-04', '2016-01-01', '2016-03-01'],
                                  'ratings': ['Rainy', 'Cloudy', 'Sunny']})

tx_weather = pd.merge_ordered(austin, houston)
print(tx_weather)
print()

tx_weather_suff = pd.merge_ordered(austin, houston, on='date', suffixes=['_aus', '_hus'])
print(tx_weather_suff)
print()

tx_weather_ffill = pd.merge_ordered(austin, houston, on='date', suffixes=['_aus', '_hus'], fill_method='ffill')
print(tx_weather_ffill)
print()
````

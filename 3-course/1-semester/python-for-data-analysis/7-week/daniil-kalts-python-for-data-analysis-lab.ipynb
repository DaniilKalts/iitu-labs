Files: `datasets/olympic-medals/bronze_top5.csv`, `datasets/olympic-medals/silver_top5.csv`

- a. If you ran `combined = bronze.append(silver)`, how many rows would **combined** have?
- b. How many rows would `combined.loc['United States']` return?

Check your answer with code. In recent pandas, `.append()` has been removed, so use `pd.concat()` or `._append()` instead.

### Exercise 4 – Concatenation with Inner Join
Continue working with the DataFrames built from **The Guardian's Olympic medal dataset**: **bronze**, **silver** and **gold** (`datasets/olympic-medals/*_top5.csv`, loaded with `index_col=0`).

Your task is to compute an inner join.

**Tasks:**
- a. Build a list of DataFrames called **medal_list** containing **bronze**, **silver** and **gold**.
- b. Concatenate **medal_list** horizontally with an inner join to create **medals**:
    - Use the keyword argument `keys=['bronze', 'silver', 'gold']` to get hierarchical indexing.
    - Use `axis=1` for horizontal concatenation.
    - Use `join='inner'` to keep only the rows whose index labels appear in all three DataFrames.

### Exercise 5. Part 5 – Use melt()
Convert the subject columns into long format:

```python
long_scores = academic_data.melt(
    id_vars=[
        "student_id", "name", "group",
        "year", "attendance"
    ],
    value_vars=["python", "statistics"],
    var_name="subject",
    value_name="score"
)
```

**Requirements:**
- a. Remove rows where `score` is missing.
- b. Capitalize the subject names.
- c. Find the mean score by subject.
- d. Find the mean score by year and subject.
- e. Find the student with the highest score in each subject.
- f. Find students whose score improved from 2025 to 2026.

````python
long_scores = academic_data.melt(
    id_vars=["student_id", "name", "group", "year", "attendance"],
    value_vars=["python", "statistics"],
    var_name="subject",
    value_name="score"
)

long_scores = long_scores.dropna(subset=["score"])
long_scores["subject"] = long_scores["subject"].str.capitalize()

print(long_scores.groupby("subject")["score"].mean())
print()

print(long_scores.groupby(["year", "subject"])["score"].mean())
print()

idx = long_scores.groupby("subject")["score"].idxmax()
top_students = long_scores.loc[idx, ["subject", "student_id", "name", "year", "score"]]
print(top_students)
print()

wide = long_scores.pivot(index=["student_id", "subject"], columns="year", values="score")
improved = wide[wide[2026] > wide[2025]]
print(improved)
````

### Exercise 5. Part 6 – Use pivot()
Transform `long_scores` back into wide format:

```python
wide_scores = long_scores.pivot(
    index=["student_id", "year"],
    columns="subject",
    values="score"
)
```

**Requirements:**
- a. Reset the index.
- b. Remove the name attached to the columns index.
- c. Compare the result with the original score columns.
- d. Explain why both `student_id` and `year` are needed in the pivot index.

**Challenge:** Try using only `student_id` as the index and explain the resulting error.

````python
wide_scores = long_scores.pivot(
    index=["student_id", "year"],
    columns="subject",
    values="score"
)

wide_scores = wide_scores.reset_index()
wide_scores.columns.name = None

print(wide_scores)
print(wide_scores.shape)

original = (
    academic_data[["student_id", "year", "python", "statistics"]]
    .dropna(subset=["python", "statistics"], how="all")
    .rename(columns={"python": "Python", "statistics": "Statistics"})
    .sort_values(["student_id", "year"])
    .reset_index(drop=True)
)
print(original.shape)
print(wide_scores.equals(original))

try:
    long_scores.pivot(index="student_id", columns="subject", values="score")
except ValueError as e:
    print("ValueError:", e)
````

**Answers:**

**d.** Each student has one row per year, so `student_id` alone doesn't identify a row uniquely. The pair (`student_id`, `year`) is unique, so `pivot()` knows exactly which cell each score belongs in.

**Challenge:** with only `student_id` as the index, pandas finds two scores for the same (student, subject) cell, one for 2025 and one for 2026. It can't put both in one cell, so it raises `ValueError: Index contains duplicate entries, cannot reshape`.
